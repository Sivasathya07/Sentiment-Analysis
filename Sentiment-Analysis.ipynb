```python
import whisper
from transformers import pipeline
from gtts import gTTS
from IPython.display import Audio

# Load Whisper Model
print("Loading Whisper Model...")
whisper_model = whisper.load_model("base")

# Speech to Text
result = whisper_model.transcribe("sample_audio.mp3")
text = result["text"]

print("\nTranscribed Text:")
print(text)

# Load DistilBERT Sentiment Model
print("\nLoading DistilBERT...")
sentiment_model = pipeline(
    "sentiment-analysis",
    model="distilbert-base-uncased-finetuned-sst-2-english"
)

# Predict Sentiment
sentiment = sentiment_model(text)[0]

label = sentiment["label"]
score = sentiment["score"]

print("\nSentiment Analysis Result")
print("Sentiment:", label)
print("Confidence:", round(score, 4))

# Create Response Text
response_text = (
    f"The sentiment of the given audio is {label}. "
    f"The confidence score is {round(score, 2)}"
)

print("\nResponse:")
print(response_text)

# Convert Response to Speech
tts = gTTS(response_text)
tts.save("output_audio.mp3")

print("\nOutput audio generated successfully!")

# Play Audio in Colab
Audio("output_audio.mp3")
```
